```python
# Membuat Dictionary (Data Mahasiswa)
siswa = {
    "nama": "Budi",
    "umur": 20,
    "jurusan": "Informatika"
}

# Access (Mengambil nilai berdasarkan Key)
print(siswa["nama"])    # Output: Budi

# Modify & Add (Mengubah data lama / menambah key baru)
siswa["umur"] = 21            # Mengubah nilai yang ada
siswa["ipk"] = 3.75           # Menambahkan key-value baru

# Delete (Menghapus key)
del siswa["jurusan"]

print("Data Siswa Akhir:", siswa)
# Output: {'nama': 'Budi', 'umur': 21, 'ipk': 3.75}
```
